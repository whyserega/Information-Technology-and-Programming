## Вопросы для подготовки

**1. Почему в Python не нужно указывать тип переменной? Что хранится в переменной?**

Python определяет тип объекта автоматически. Имя переменной ссылается на объект со значением, у которого уже есть тип. Например, после `a = 7` объект имеет тип `int`, а после `a = "текст"` имя `a` ссылается на строку.

**2. Чем отличаются `int` и `float`? Когда деление даёт `float`, а когда `int`?**

`int` используется для целых чисел, а `float` — для вещественных (хранятся с ограниченной точностью). Обычное деление `/` даёт `float`, даже при `10 / 2`. Оператор `//` при делении двух `int` возвращает `int`, например `10 // 3 == 3`.

**3. Чем отличается `print()` с несколькими аргументами от конкатенации? Для чего f-строки?**

В `print("Возраст:", 19)` аргументы разделяются пробелом, и число отдельно преобразовывать не надо. При сложении строк через `+` обе части должны быть строками: `"Возраст: " + str(19)`. F-строки удобны для вставки переменных: `f"Возраст: {19}"`.

**4. Для чего нужна функция `type()`? Почему типы `5` и `5.0` различаются?**

`type(x)` показывает тип объекта. У `5` тип `<class 'int'>`, а у `5.0` — `<class 'float'>`, потому что первое значение записано как целое число, а второе — как вещественное.

**5. Можно ли сначала сохранить в переменной число, затем строку?**

Да: `value = 10`, а затем `value = "десять"`. Это свойство динамической типизации. Однако если часто менять тип переменной, можно запутаться и получить ошибку в вычислениях.

**6. Почему `int("abc")` вызывает ошибку, а `int("123")` — нет?**

`"123"` — строка с корректной записью целого числа, а `"abc"` не может быть превращена в число. Для защиты от ошибки используют `try` и `except ValueError`.

## Контрольные вопросы

**1. Что такое переменная в Python? Примеры правильных и неправильных имён.**

Переменная — это имя, через которое обращаются к значению. Присваивание записывают через `=`: `age = 19`, `user_name = "Анна"`, `total2 = 50`. Имя `2total` недопустимо, потому что начинается с цифры.

**2. Назовите минимум четыре простых типа данных и объясните динамическую типизацию.**

`int`: `5` → `<class 'int'>`; `float`: `5.0` → `<class 'float'>`; `str`: `"Привет"` → `<class 'str'>`; `bool`: `True` → `<class 'bool'>`. В Python тип принадлежит объекту, а имя переменной можно связать с объектом другого типа.

**3. Как работает `print()`? Приведите три примера.**

```python
x = 5
print(x)                 # 5
print("Число:", x)       # Число: 5
print(f"Число: {x}")     # Число: 5
```
Первый вариант выводит одно значение, второй — несколько аргументов через пробел, третий — готовую форматированную строку.

**4. Что делает `type()`? Покажите типы числа, строки и логического значения.**

```python
a = 12
s = "дом"
flag = True
print(type(a))     # <class 'int'>
print(type(s))     # <class 'str'>
print(type(flag))  # <class 'bool'>
```
Функция возвращает класс (тип) переданного объекта.

**5. Какие арифметические операции поддерживает Python?**

`+` — сложение (`5 + 2.0` → `7.0`, `float`); `-` — вычитание (`5 - 2` → `3`, `int`); `*` — умножение (`5 * 2` → `10`, `int`); `/` — обычное деление (`5 / 2` → `2.5`, `float`); `//` — деление с округлением вниз (`5 // 2` → `2`, `int`); `%` — остаток (`5 % 2` → `1`, `int`); `**` — степень (`5 ** 2` → `25`, `int`). С вещественными операндами многие результаты будут `float`.

**6. В чём разница между `/` и `//`, в том числе для отрицательных чисел?**

`/` выполняет обычное деление и возвращает `float`: `7 / 2 = 3.5`, `-7 / 2 = -3.5`. Оператор `//` округляет частное вниз: `7 // 2 = 3`, `-7 // 2 = -4`. Аналогично `7 // -2 = -4`. При двух целочисленных операндах результат `//` — `int`.

**7. Как выполняется преобразование типов? Когда возникают ошибки?**

```python
int("10")     # 10
int(3.9)      # 3
float("2.5")  # 2.5
str(15)       # '15'
```
При `int("abc")` возникнет `ValueError`, потому что строка не содержит подходящего числа.

**8. Что значит, что выражение возвращает значение?**

Выражение вычисляется, после чего результат можно записать в переменную или напечатать: `a = 2 + 3` (число `5`), `b = 7 > 3` (`True`), `c = True and False` (`False`). Можно также написать `print(2 + 3)`.

**9. Как работают сравнения и логические операторы? Каков их приоритет?**

`==`, `!=`, `<`, `>`, `<=`, `>=` сравнивают значения и дают `True` или `False`. Операторы `and`, `or`, `not` объединяют или изменяют условия. Например, `(5 > 2) and (3 < 4)` → `True`; `(5 < 2) or (3 == 3)` → `True`; `not (7 == 7)` → `False`. Сначала выполняются сравнения, затем `not`, потом `and`, затем `or`; скобки могут менять порядок.

**10. Что будет при операциях над объектами разных типов?**

`5 + 2.5` даёт `7.5` типа `float`: число `int` неявно преобразуется для вычисления. А `"5" + 2` вызовет `TypeError`, поскольку строка с числом автоматически не складывается. Нужно написать `int("5") + 2` либо `"5" + str(2)`.

**11. Какими способами можно вывести несколько значений через `print()`?**

```python
name, age = "Максим", 21
print(name, age)                      # аргументы, через пробел
print(f"{name}: {age}")               # f-строка
print("{}: {}".format(name, age))      # format()
print(name + ": " + str(age))         # конкатенация
```
Для простого вывода подходит запятая, для сообщений с переменными обычно удобнее f-строка.

**12. Что такое порядок операций и как на него влияют скобки?**

Сначала выполняются скобки и возведение в степень, потом умножение и деление, затем сложение и вычитание. `2 + 3 * 4` даст `14`, а `(2 + 3) * 4` даст `20`. Скобки позволяют явно задать нужный порядок.

**13. Как устроено множественное присваивание?**

```python
a, b = 1, 2       # два значения за одно присваивание
a, b = b, a       # обмен без дополнительной переменной
x, y = [10, 20]   # распаковка списка
```
Количество имён должно совпадать с количеством распаковываемых значений (если не используется `*`).

**14. Какие типы называются изменяемыми и неизменяемыми?**

`int`, `float`, `bool`, `str` — неизменяемые: значение существующего объекта нельзя менять «на месте». Например, `word = "кот"`, затем `word = word + "ик"` создаёт новую строку и меняет ссылку `word`. А инструкция `word[0] = "р"` вызовет `TypeError`. Списки, напротив, изменяемые.
