Ниже приведены собственные короткие примеры по пяти перечисленным темам: проверка диапазона, сумма через `for`, счётчик `while`, цикл с `break/else`, квадратный корень с проверкой.

````python
# Примеры для подготовки к практическому занятию №2
import math

# 1. Простой if/else
x = 7
if 0 <= x <= 10:
    print("1. Число в диапазоне от 0 до 10")
else:
    print("1. Число вне диапазона")

# 2. Нахождение суммы циклом for
total = 0
for i in range(1, 6):
    total += i
print("2. Сумма чисел от 1 до 5:", total)

# 3. Цикл while со счётчиком
count = 1
while count <= 3:
    print("3. Счётчик:", count)
    count += 1

# 4. Блок else не сработает из-за break
for num in range(5):
    if num == 3:
        print("4. Найдено число 3, выход из цикла")
        break
else:
    print("4. Цикл завершён без break")

# 5. Корень с проверкой
value = -4
if value < 0:
    print("5. Нельзя вычислить вещественный корень")
else:
    print("5. Корень:", math.sqrt(value))
````

## Вопросы для обсуждения

**1. В каких случаях лучше использовать `for`, а в каких — `while`?**

`for` удобен, когда известно количество повторений: например, вывести таблицу умножения от 1 до 10. `while` лучше, когда цикл продолжается до события или условия: например, ввод пароля до правильного ответа.

**2. Как работает `else` у цикла и почему его легко перепутать?**

Блок `else` после цикла выполняется при обычном завершении, но не после `break`. Это не означает «если условие неверно» в привычном смысле, поэтому такой блок легко понять неправильно.

**3. Какие ошибки часто встречаются в циклах?**

Например, в `while` забывают обновить счётчик — получится бесконечный цикл. При `range(1, 10)` число 10 не выводится, поскольку верхняя граница не включается. Ещё одна ошибка — проверять не то условие выхода.

**4. Чем встроенные `sum`, `min`, `max` удобнее ручного подсчёта?**

Встроенные функции делают код короче и понятнее: `sum([1, 2, 3])` вместо отдельного цикла. Цикл полезен, когда по ходу подсчёта нужна особая проверка или другая логика.

**5. Как правильно пользоваться функциями `math`?**

Перед вызовом нужно написать `import math` и обращаться как `math.sqrt(9)`. Вызов `sqrt(9)` без отдельного импорта приведёт к `NameError`, а `math.sqrt(-9)` — к `ValueError` для вещественных чисел.

**6. Как обрабатывать деление на ноль и отрицательное под корнем?**

Перед вычислением проверяют входные данные через `if/elif/else`. Например, если знаменатель равен нулю, выводят сообщение вместо деления; корень вычисляют только при неотрицательном аргументе.

**7. Как контролировать типы данных после `input()`?**

`input()` возвращает строку, поэтому `"5" + "2"` даёт `"52"`, а не `7`. Для вычислений ввод преобразуют через `int()` или `float()`, а ошибку конвертации можно обработать через `try/except ValueError`.

**8. Как выбрать `pow`, `**` или `math.factorial`?**

Оператор `**` удобен для обычного возведения в степень: `2 ** 3`. `math.pow(2, 3)` возвращает `float`, а `math.factorial(5)` вычисляет `5!` для неотрицательного целого числа. Цикл для факториала полезен в учебной задаче, где требуется показать алгоритм.

**9. Как проверять результат приближённых вычислений?**

Проверяют разницу между соседними приближениями, задают максимально допустимое количество шагов и сравнивают с точным значением, если оно доступно, например с `math.sqrt(x)`. Для вещественных чисел лучше сравнивать погрешность, а не требовать строгого равенства.

### Уровень 1. Базовый

**Задание:** пользователь вводит целое число. Определить, делится ли оно одновременно на 3 и 5, только на 3, только на 5 или ни на одно из них. Применить `if/elif/else`.

Примеры: `15` — на оба числа; `9` — только на 3; `10` — только на 5; `7` — ни на одно.

````python
# Вариант 9. Базовый уровень
try:
    number = int(input("Введите целое число: "))

    if number % 3 == 0 and number % 5 == 0:
        print("Число делится и на 3, и на 5")
    elif number % 3 == 0:
        print("Число делится только на 3")
    elif number % 5 == 0:
        print("Число делится только на 5")
    else:
        print("Число не делится ни на 3, ни на 5")
except ValueError:
    print("Ошибка: нужно ввести целое число")
````

### Уровень 2. Средний

**Задание:** пользователь вводит число `x` (в радианах). Вычислить `math.sin(x)`, `math.cos(x)`, `math.tan(x)` с точностью до трёх знаков и использовать `math.isclose` для проверки случаев, когда тангенс не определён.

**Уточнение:** тангенс не определён **при нечётных кратных `π/2`** (`π/2`, `3π/2`, ...). При чётных кратных `π/2`, например при `π`, тангенс существует и равен нулю. Для проверки используется приближённое сравнение из-за неточности чисел `float`.

````python
# Вариант 9. Средний уровень
import math

try:
    x = float(input("Введите x в радианах: "))

    if not math.isfinite(x):
        print("Ошибка: нужно ввести конечное число")
    else:
        print(f"sin({x}) = {math.sin(x):.3f}")
        print(f"cos({x}) = {math.cos(x):.3f}")

        k = round(x / (math.pi / 2))
        near_multiple = math.isclose(x, k * math.pi / 2,
                                     rel_tol=1e-12, abs_tol=1e-12)

        if near_multiple and k % 2 != 0:
            print("Предупреждение: тангенс в этой точке не определён")
        else:
            print(f"tan({x}) = {math.tan(x):.3f}")
except ValueError:
    print("Ошибка: нужно ввести число")
````

### Уровень 3. Продвинутый

**Задание:** пользователь вводит число `n` до 100. Построить треугольник Паскаля до `n`-й строки, используя списки и циклы `for`; числа вычислять через сочетания `math.comb()` либо вручную. Вывести треугольник с выравниванием по центру.

Считаем строки **с нулевой**, поэтому при `n = 5` получаются строки с номерами от 0 до 5 включительно. Значения вычисляются по формуле `C(n, k) = n! / (k! · (n − k)!)`.

````python
# Вариант 9. Продвинутый уровень
import math

try:
    n = int(input("Введите номер последней строки треугольника (0-100): "))

    if not 0 <= n <= 100:
        print("Ошибка: число n должно быть от 0 до 100")
    else:
        triangle = []

        for i in range(n + 1):
            row = []
            for j in range(i + 1):
                row.append(math.comb(i, j))
            triangle.append(row)

        cell_width = len(str(math.comb(n, n // 2))) + 1
        formatted_rows = []
        for row in triangle:
            row_text = " ".join(f"{value:^{cell_width}}" for value in row)
            formatted_rows.append(row_text)

        total_width = len(formatted_rows[-1])
        print(f"Треугольник Паскаля до строки {n}:")
        for row_text in formatted_rows:
            print(row_text.center(total_width))
except ValueError:
    print("Ошибка: нужно ввести целое число")
````

- **Средний уровень:** `0` → `sin = 0.000`, `cos = 1.000`, `tan = 0.000`; при `x = 1.5707963267948966` появится предупреждение о тангенсе.

## Контрольные вопросы

В методичке первый пункт раздела содержит лишь указание предоставить ответы в Python Notebook. Ниже даны ответы на **15 содержательных вопросов**, идущих далее.

**1. Чем принципиально отличаются циклы `for` и `while`?**

`for` обычно перебирает элементы или заданный диапазон. Например, `for i in range(5)` выводит пять чисел. `while` повторяется, пока условие истинно: например, `while password != "1234"` при проверке ввода.

**2. Когда выполняется `else` у циклов? Примеры с `break` и без него.**

```python
for i in range(3):
    print(i)
else:
    print("Цикл завершился")  # выведется

for i in range(3):
    if i == 1:
        break
else:
    print("Цикл завершился")  # не выведется
```
`else` выполняется, если цикл закончился без `break`.

**3. Чем отличаются `break` и `continue`? Пример с `while`.**

`continue` пропускает оставшийся код текущей итерации, а `break` завершает весь цикл.
```python
i = 0
while i < 5:
    i += 1
    if i == 2:
        continue  # 2 пропускается
    if i == 4:
        break     # цикл останавливается
    print(i)      # выведет 1 и 3
```

**4. Назовите минимум три типичные ошибки в циклах и их исправления.**

1. **Не обновляется счётчик:** `i = 0; while i < 3: print(i)` — бесконечный цикл. Исправление: добавить `i += 1` внутри цикла.
2. **Неверный диапазон:** `range(1, 10)` для чисел от 1 до 10 пропускает 10. Исправление: `range(1, 11)`.
3. **Неправильное условие:** `while n < 0:` не выполнится для положительного `n`, когда нужно уменьшать его до нуля. Исправление: `while n > 0:` и `n -= 1`.

**5. Как устроено `if/elif/else`? Почему важен порядок?**

```python
x = 8
if x < 0:
    print("Меньше нуля")
elif x <= 10:
    print("От 0 до 10")
else:
    print("Больше 10")
```
Проверки идут сверху вниз, и выполняется только первая подходящая ветка. Поэтому более узкие условия иногда нужно ставить раньше общих.

**6. Что делают `and`, `or`, `not`? Составьте сложное условие.**

`and` требует выполнения обоих условий, `or` — хотя бы одного, `not` — обращает логическое значение. Заданное условие:
```python
(n % 2 == 0 and 10 <= n <= 20) or (n % 2 != 0 and n > 100)
```
Оно истинно для чётных чисел от 10 до 20 или нечётных чисел больше 100.

**7. Каков приоритет арифметических операций? Приведите примеры.**

Сначала вычисляются скобки, затем `**`, затем `*`, `/`, `//`, `%`, потом `+` и `-`.

- `2 + 3 * 4` → `14`;
- `(2 + 3) ** 2` → `25`;
- `17 // 5 + 17 % 5` → `5` (так как `3 + 2 = 5`).

**8. Чем отличаются `/` и `//`? Какие у них типы?**

```python
a, b = 7, 2
print(a / b, type(a / b))    # 3.5, float
print(a // b, type(a // b))  # 3, int
```
`/` выполняет обычное деление; `//` округляет частное вниз. Для `float`-операндов `//` тоже может вернуть `float`.

**9. Как работают `min`, `max`, `sum`, `abs`, `round`?**

`min([3, 1, 2])` → `1` (минимум последовательности чисел); `max([3, 1, 2])` → `3` (максимум); `sum([3, 1, 2])` → `6` (сумма); `abs(-5)` → `5` (модуль числа); `round(3.14159, 2)` → `3.14` (округление `float`). Для `min`, `max`, `sum` обычно передают коллекцию совместимых значений, для `abs`, `round` — число.

**10. Для чего нужна `range(start, stop, step)`?**

Она создаёт диапазон целых чисел; `start` — начало, `stop` — невключённая граница, `step` — шаг.
```python
list(range(1, 5))      # [1, 2, 3, 4]
list(range(5, 0, -1))   # [5, 4, 3, 2, 1]
list(range(0, 10, 3))   # [0, 3, 6, 9]
```
Последний аргумент не может быть равен нулю.

**11. Как импортировать `math` и какие функции в нём полезны?**

После `import math` доступны, например, `math.sqrt(x)` (корень), `math.factorial(n)` (факториал), `math.sin(x)` (синус), `math.cos(x)` (косинус), `math.ceil(x)` (округление вверх). Правильно: `math.sqrt(16)`. Ошибка: написать `sqrt(16)` без соответствующего импорта.

**12. Как найти квадратный корень тремя способами?**

`math.sqrt(9)` → `3.0`: удобный готовый способ для неотрицательных чисел. `9 ** 0.5` → `3.0`: короткая запись, но с отрицательными числами результат может стать комплексным. **15. Как выбрать между циклом и встроенной функцией?**
