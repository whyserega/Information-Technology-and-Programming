## Задание

Текстовый файл `pr6.txt` содержит строку из заглавных латинских букв и точек, всего не более чем из $10^6$ символов. Определите максимальное количество идущих подряд символов, среди которых нет точек, а количество гласных (букв A, E, I, O, U, Y) не превышает 7.
Например, для строки символов:
`ASD.DFGUIO.OOOAEEERUY.UYTEE`
под условие задачи подходя подстроки: `ASD`, `DFGUIO` и `UYTEE`. Среди допустимых последовательностей максимальную длину имеет подстрока `DFGUIO`. Ответ: 6.



## Решение

Файл pr6.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Реализуем чтение строки из файла

````python
mystr = open('pr6.txt').read().strip()

# Посмотрим, что мы считали...
print(len(mystr))
print(mystr[:50])
````

Теперь ответим на поставленные вопросы, используя строку `mystr`

````python
# Считываем исходную строку из файла
with open('pr6.txt', encoding='utf-8') as f:
    mystr = f.read().strip()

# Максимум 7 гласных, точка не должна входить в найденную подстроку
vowels = 'AEIOUY'
max_len = 0

# Сначала делим исходную строку по точкам
for part in mystr.split('.'):
    left = 0
    vowel_count = 0

    for right in range(len(part)):
        if part[right] in vowels:
            vowel_count += 1

        # Сдвигаем левую границу, если гласных стало больше семи
        while vowel_count > 7:
            if part[left] in vowels:
                vowel_count -= 1
            left += 1

        current_len = right - left + 1
        if current_len > max_len:
            max_len = current_len

print(f'Максимальная подходящая подстрока: {max_len}')
````

## Индивидуальное задание — вариант 9

Текстовый файл `pr6_1.txt` содержит заглавные латинские буквы и цифры. Нужно определить, **сколько раз буква `L` стоит сразу после цифры**.

````python
from pathlib import Path
from urllib.request import urlretrieve

# Скачиваем данные только в том случае, если файла ещё нет в Colab
file_name = 'pr6_1.txt'
if not Path(file_name).exists():
    url = 'https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/pract6/pr6_1.txt'
    urlretrieve(url, file_name)

with open(file_name, encoding='utf-8') as f:
    text = f.read().strip()

# Проверяем пары соседних символов
count = 0
for i in range(1, len(text)):
    if text[i] == 'L' and text[i - 1].isdigit():
        count += 1

print('Количество букв L сразу после цифры:', count)
````
