## Анализ задания, получение данных

Файл data_v3.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Реализуем чтение всего файла построчно. Используем метод `open()`, а затем будем обходить файл как коллекцию с использованием цикла `for line in File`. Очевидно, что csv-файл используется в данной работе как обычный текстовый файл и в цикле `for` производится последовательный обход всех строк этого файла, то есть текстовый файл - это коллекция его строк.
В файле много строк, поэтому будем выводить только первые `n_to_print` строк, но обрабатывать будем все (чтобы не возникло скрытых ошибок)

````python
F=open('data_v3.csv')
n_to_print=10
for line in F:
  n_to_print-=1
  if n_to_print>0:
    print(line)
````

Строки перебираются, но теперь необходимо разделить единую строку на отдельные элементы данных. Для этого выполним метод `split(sep)`, который разрезает строку, используя символ `sep` в качестве разделителя. Также обратим внимание на тот факт, что в первой строке содержатся не данные, а заголовки столбцов.

````python
# доработка предыдущего кода
F=open('data_v3.csv')
header = F.readline().strip().split(',')
print(header)
print('-'*20)
n_to_print=10
for line in F:
  # производим обработку в любом случае ...
  data = line.strip().split(',')
  # ... аа печатаем только первые n_to_print
  n_to_print-=1
  if n_to_print>0:
    print(data)
````

Метод `split()` вернул список отдельных строк - полстрок исходной строки. Теперь необходимо сформировать структуру данных максимально полно описывающую предметную область.
Для этого сформируем список словарей, то есть все данные в процессе чтения будем построчно перемещать в список `all_clients`, при этом каждый клиент будет представлен в виде словаря. Например первая строка в файле будет выглядеть следующим образом:

`example = {'ID' : '12-9321320', 'LastName':'Pickle', 'FirstName':'Réjane', 'email':None, 'Plus':41564.80, 'Minus':46160.22, 'credit_card':'3557725016917448', 'Tax':16408.20}`

то есть будет сформирована структура данных в которой все элементыы представляют собой пары вида `ключ` : `значение`, при этом как и в списке возможно обращение к элементамм данной структуры по индексу, только в словарях вместо целочисленного индекса используется ключ.
Например, применительно к первой строке, обратиться к сумме поступления для ее печати можно следующим образом:

`print( example['Plus'] )`

````python
# Доработаем предыдущий скрипт
F=open('data_v3.csv')
header = F.readline().strip().split(',')

# список всех клиентов
all_clients=[]

for line in F:
  data = line.strip().split(',')
  client = {} # создаем пустой словарь
  for i, key in enumerate(header): # обходим все заголовки
    client[key] = data[i] # добавляем в словарь клюс:значение

  # некоторые значения придется конвертировать к подходящим типам
  client['Minus'] = float(client['Minus'][1:])
  client['Plus'] = float(client['Plus'][1:])
  client['Tax'] = float(client['Tax'][1:])

  all_clients.append(client) # добавляем клиента в скписок

print(all_clients[:10]) # напечатаем первых 10 клиентов

F.close() # файл больше не нужен
````

## Решение задачи 1

Первая задача:
Найдите держателя карты с максимальной прибылью. В ответе необходимо указать ФИО держателя и максимальную прибыль (прибыль = поступления – списания).

Переберем все элементы в списке `all_clients` и выберем клиента с максимальной разницей `client['Plus'] - client['Minus']`



````python
# переменные для хранения целевого клиента и максимальной прибыли
target_client=[]
max_profit=0

for client in all_clients:
  if client['Plus'] - client['Minus'] > max_profit:
    max_profit = client['Plus'] - client['Minus']
    target_client = client

print('Максимальная прибыль: ', max_profit)
print('Клиент с максимальной прибылью', target_client)
````

## Решение задачи 2

Вторая задача: Найдите количество держателей карт, у которых сумма уплаченного налога больше среднего арифметического значения налога по всей выборке.

Задачу можно решить также перебором всей выборки, но предварительно необходимо определить среднее значение уплаченных налогов

````python
# Определим среднее значение нолога
total_tax = 0
count_tax = len(all_clients)

for client in all_clients:
  total_tax += client['Tax']

avg_tax = total_tax / count_tax

print('Среднее значение налога = %.2f' % avg_tax)

counter=0
for client in all_clients:
  if client['Tax'] > avg_tax:
    counter+=1

print('Количество клиентов с налогом выше среднего: ', counter)
````

В данной работе были использованы операторы `if` (условный оператор), `for ... in range(...)` (оператор цикла с обходом диапазона). Применялись функции `print()`, `len()`, `max()`. В работе использовался тип данных `dict()`

## Индивидуальное задание — вариант 9

В файле `p7_data_09.csv` находятся сведения о сотрудниках компании, их зарплатах и налогах. Требуется определить:

1. У кого минимальная зарплата и кто получает меньше всех после уплаты налога.
2. Сколько сотрудников не имеют указанного навыка (`skill`).
3. Среднюю зарплату отдельно для сотрудников с e-mail и без e-mail.
4. Общую сумму налогов по всей выборке.

````python
from pathlib import Path
from urllib.request import urlretrieve
import csv

file_name = 'p7_data_09.csv'
if not Path(file_name).exists():
    url = 'https://raw.githubusercontent.com/enikolaev/IT-and-Programming/main/data/pract7/p7_data_09.csv'
    urlretrieve(url, file_name)

# Читаем таблицу: каждая строка становится словарём
with open(file_name, newline='', encoding='utf-8-sig') as f:
    reader = csv.DictReader(f)
    employees = list(reader)
    columns = reader.fieldnames

print('Количество сотрудников:', len(employees))
print('Столбцы таблицы:', columns)
print('Первая запись:', employees[0] if employees else 'нет данных')
````

Теперь найдём нужные поля в таблице и преобразуем денежные суммы в числа. Для отсутствующего e-mail или навыка будем учитывать пустые значения.

````python
# Названия полей могут немного отличаться, поэтому найдём их по заголовкам

def get_column(names):
    for col in columns:
        name = col.lower().replace('_', '').replace('-', '').replace(' ', '')
        if name in names:
            return col
    raise ValueError(f'В таблице не найдено поле: {names}. Есть столбцы: {columns}')

salary_col = get_column(('salary', 'wage', 'pay', 'зарплата', 'зп'))
tax_col = get_column(('tax', 'taxes', 'налог'))
skill_col = get_column(('skill', 'skills', 'навык'))
email_col = get_column(('email', 'mail', 'почта'))

# Определим имя сотрудника по полям таблицы

def employee_name(row):
    fields = {key.lower().replace('_', '').replace('-', '').replace(' ', ''): key for key in row}
    for name_key in ('fullname', 'fio', 'фио', 'employee', 'employee_name'):
        name_key = name_key.replace('_', '')
        if name_key in fields and row[fields[name_key]]:
            return row[fields[name_key]]
    first = next((row[fields[k]] for k in ('firstname', 'first', 'имя', 'name') if k in fields), '')
    last = next((row[fields[k]] for k in ('lastname', 'surname', 'last', 'фамилия') if k in fields), '')
    return (str(last) + ' ' + str(first)).strip() or str(row.get('ID', row))


def to_number(value):
    return float(str(value or '0').replace('$', '').replace(' ', '').replace(',', '') or 0)

for emp in employees:
    emp['salary_number'] = to_number(emp[salary_col])
    emp['tax_number'] = to_number(emp[tax_col])

print('Данные подготовлены.')
````

### Задание 1. Минимальная зарплата и минимальная сумма после налогов

````python
min_salary_emp = min(employees, key=lambda emp: emp['salary_number'])
min_net_emp = min(employees, key=lambda emp: emp['salary_number'] - emp['tax_number'])

print('Сотрудник с минимальной зарплатой:', employee_name(min_salary_emp))
print(f"Зарплата: {min_salary_emp['salary_number']:.2f}")
print()
print('Сотрудник с минимальной зарплатой после налога:', employee_name(min_net_emp))
print(f"После налога: {min_net_emp['salary_number'] - min_net_emp['tax_number']:.2f}")
````

### Задание 2. Количество сотрудников без навыка (`skill`)

````python
no_skill = 0
for emp in employees:
    if str(emp[skill_col] or '').strip().lower() in ('', '-', 'none', 'null', 'nan'):
        no_skill += 1

print('Сотрудников без указанного навыка:', no_skill)
````

### Задание 3. Средняя зарплата с e-mail и без e-mail

````python
with_email = []
without_email = []

for emp in employees:
    if str(emp[email_col] or '').strip().lower() not in ('', '-', 'none', 'null', 'nan'):
        with_email.append(emp['salary_number'])
    else:
        without_email.append(emp['salary_number'])

if with_email:
    print(f'Средняя зарплата сотрудников с e-mail: {sum(with_email) / len(with_email):.2f}')
else:
    print('Сотрудников с e-mail нет')

if without_email:
    print(f'Средняя зарплата сотрудников без e-mail: {sum(without_email) / len(without_email):.2f}')
else:
    print('Сотрудников без e-mail нет')
````

### Задание 4. Общая сумма налогов

````python
total_tax = sum(emp['tax_number'] for emp in employees)
print(f'Сумма налога по всей выборке: {total_tax:.2f}')
````

### 14. В чём разница между `pop()` и `popitem()` у словаря?
