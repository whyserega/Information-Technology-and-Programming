## Задание

В файле pr5.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от –10 000 до 10 000 включительно. Определите количество пар, в которых оба элемента больше, чем среднее арифметическое всех чисел в файле, а их сумма оканчивается на 31. В ответе запишите два числа: сначала количество найденных пар, а затем – максимальную сумму элементов таких пар. В данной задаче под парой подразумевается два идущих подряд элемента последовательности. При решении задачи необходимо использовать списочный тип данных.


## Решение

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5.txt
````

Можно скопировать и поместить файл с данными любым удобным способом

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

````python
# Создаем пустой список
data = []
# Выполняем операции чтения и заполнение списка
with open('pr5.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

# Посмотрим, что мы считали...
print(len(data))
print(data[:10])
````

Теперь ответим на поставленные вопросы использую список `data`

````python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
````

````python
# Подсчет количества подходящих пар
counter = 0
max_sum = -10000
for i in range(len(data)-1):
  # пара элементов
  x, y = data[i], data[i+1]
  # проверяем, подходит ли пара
  if x>sr and y>sr and str(x+y)[-2:] == '31':
    counter +=1
    max_sum = max(max_sum, x+y)
    # print(x, y, x+y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
````

# Индивидуальное задание
**Вариант 9**

В файле pr5_1.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
0 до 10 000 включительно. Определите количество пар, в которых
хотя бы один из двух элементов меньше, чем среднее
арифметическое всех чисел в файле, и хотя бы один из двух
элементов делится на 7 и не делится на 3, 11 и 13. В ответе запишите
два числа: сначала количество найденных пар, а затем –
минимальную сумму элементов таких пар. В данной задаче под
парой подразумевается два идущих подряд элемента
последовательности

````python
# Скачиваем файл с исходными данными
!wget -q -O pr5_1.txt https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_1.txt

# Считываем все числа из файла в список
data1 = []
with open('pr5_1.txt') as F:
    for line in F:
        data1.append(int(line))

# Находим среднее арифметическое всех чисел
sr = sum(data1) / len(data1)

# Ищем подходящие пары соседних элементов
counter = 0
min_sum = 20001

for i in range(len(data1) - 1):
    x, y = data1[i], data1[i + 1]

    # Хотя бы один элемент меньше среднего арифметического
    condition1 = x < sr or y < sr

    # Хотя бы один элемент делится на 7 и не делится на 3, 11 и 13
    condition2 = ((x % 7 == 0 and x % 3 != 0 and x % 11 != 0 and x % 13 != 0) or
                  (y % 7 == 0 and y % 3 != 0 and y % 11 != 0 and y % 13 != 0))

    if condition1 and condition2:
        counter += 1
        min_sum = min(min_sum, x + y)

print(f'Количество подходящих пар = {counter}')
print(f'Минимальная сумма = {min_sum}')
````

## Ответы на контрольные вопросы

### 1. Что покажет приведённый фрагмент кода?

```python
numbers = [1, 2, 3, 4, 5]
numbers[2] = 99
print(numbers)
```

**Ответ:** `[1, 2, 99, 4, 5]`

Индекс `2` соответствует третьему элементу списка, поэтому число `3` заменяется на `99`.

---

### 2. Что покажет приведённый фрагмент кода?

```python
numbers = list(range(3))
print(numbers)
```

**Ответ:** `[0, 1, 2]`

`range(3)` создаёт последовательность чисел от `0` до `2`, а `list()` преобразует её в список.

---

### 3. Что покажет приведённый фрагмент кода?

```python
numbers = [10] * 5
print(numbers)
```

**Ответ:** `[10, 10, 10, 10, 10]`

При умножении списка на `5` его содержимое повторяется пять раз.

### 4. Как добавить строку `'Вэнди'` в пустой список `names = []` на позицию с индексом 0?

Правильная инструкция:

```python
names.append('Вэнди')
```

После выполнения список будет иметь вид: `['Вэнди']`.

Метод `append()` добавляет элемент в конец списка. Так как список изначально пустой, новый элемент получает индекс `0`.

Вариант `names(0) = 'Вэнди'` неверный: круглые скобки используются для вызова функций. Для обращения к элементу списка применяются квадратные скобки, но запись `names[0] = 'Вэнди'` для пустого списка тоже вызовет ошибку, потому что элемента с индексом `0` ещё нет.

### 5. Списковые методы

**а) `index()`** — возвращает индекс первого элемента с указанным значением.

```python
numbers = [5, 10, 15]
numbers.index(10)
```

Результат: `1`.

**б) `insert()`** — вставляет элемент в список на указанную позицию.

```python
numbers = [1, 3]
numbers.insert(1, 2)
```

После выполнения: `[1, 2, 3]`.

**в) `sort()`** — сортирует элементы списка. По умолчанию сортировка выполняется по возрастанию.

```python
numbers = [3, 1, 2]
numbers.sort()
```

После выполнения: `[1, 2, 3]`.

**г) `reverse()`** — изменяет порядок элементов списка на обратный.

```python
numbers = [1, 2, 3]
numbers.reverse()
```

После выполнения: `[3, 2, 1]`.
