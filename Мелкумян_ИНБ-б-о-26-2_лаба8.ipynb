````python
import itertools

# Посмотрим краткую справку по функции перестановок
print(itertools.permutations.__doc__.splitlines()[0])
````

````python
import itertools
# допустимый алфавит
alf = 'СПОРТЛОТО'
number = 0
for word in sorted(set(itertools.permutations(alf))):
  word = ''.join(word)
  if (word[0] == 'О' and word[-1] != 'О') \
  or (word[0] != 'О' and word[-1] == 'О'):
    number += 1
    # print(number, word)
print(number)
````

## Индивидуальное задание — вариант 9

Григорий составляет 16-буквенные слова из букв слова `АНТИУТОПИЯ`. Каждая буква может встречаться несколько раз или не встречаться совсем. Требуется найти количество слов, содержащих подряд **`АНТИУТОПИЯ`**, причём **остальные гласные** идут в алфавитном порядке, а **остальные согласные** — в обратном алфавитном порядке.

Длина обязательной комбинации — 10, значит остаётся заполнить 6 позиций. Будем перебирать варианты для этих шести букв при помощи модуля `itertools`.

````python
from itertools import product

word = 'АНТИУТОПИЯ'
letters = sorted(set(word))
vowels = set('АИОУЯ')
free_places = 16 - len(word)

# Посчитаем подходящие последовательности из 6 дополнительных букв
suitable = 0
for extra in product(letters, repeat=free_places):
    vowel_part = [c for c in extra if c in vowels]
    consonant_part = [c for c in extra if c not in vowels]

    if vowel_part == sorted(vowel_part) and consonant_part == sorted(consonant_part, reverse=True):
        suitable += 1

# Комбинация АНТИУТОПИЯ может начинаться на любой из 7 позиций
positions = free_places + 1
answer = suitable * positions

print('Подходящих вариантов дополнительных букв:', suitable)
print('Возможных позиций обязательной комбинации:', positions)
print('Количество слов:', answer)
````

## Ответы на контрольные вопросы

### 1. Что такое модуль в Python?

Модуль — файл с кодом Python, в котором могут находиться функции, переменные и классы. Его можно подключить к другой программе и использовать готовый код.

---

### 2. Сколько модулей может подключить программист?

Можно подключить столько модулей, сколько нужно для программы, если они доступны в среде Python. Строгого небольшого ограничения на количество нет.

---

### 3. Как подключить модуль?

Используется команда `import`. Например:

```python
import math
print(math.sqrt(25))  # 5.0
```

Можно импортировать отдельную функцию: `from math import sqrt`.

---

### 4. Основные библиотеки Python и их назначение

- `math` — математические функции: корень, факториал, тригонометрия.
- `random` — получение случайных чисел и случайный выбор элементов.
- `datetime` — работа с датами и временем.
- `os` — работа с операционной системой, папками и файлами.
- `csv` — чтение и запись таблиц CSV.
- `itertools` — перебор комбинаций, перестановок и других последовательностей.
- `collections` — дополнительные структуры данных, например `Counter`.

---

### 5. Какие функции и классы есть в `itertools`?

- `permutations()` — все перестановки элементов.
- `combinations()` — сочетания без повторений.
- `combinations_with_replacement()` — сочетания с повторениями.
- `product()` — все комбинации элементов по позициям (декартово произведение).
- `count()` — бесконечная последовательность чисел с заданным шагом.
- `repeat()` — повторение одного значения.
- `chain()` — объединение нескольких последовательностей при переборе.

Большинство из этих средств возвращают итераторы, позволяющие получать варианты по одному, не создавая сразу огромный список.
