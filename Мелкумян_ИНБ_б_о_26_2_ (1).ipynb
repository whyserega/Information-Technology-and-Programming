## Задание

Необходимо вычислить значение выражения:
$$g=w\cdot h+b\cdot q3-q3+y/w$$

Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $w, h, b, q3, y|w$. Переменная, которая должна быть вычислена на основе независимых: $g$.


```python
w=10
h=2.5
b=3
q3=4
y_w=6.7
print("w = ", w, ", type(w) = ", type(w))
print("h = ", h, ", type(h) = ", type(h))
print("b = ", b, ", type(b) = ", type(b))
print("q3 = ", q3, ", type(q3) = ", type(q3))
print("y|w = ", y_w, ", type(y|w) = ", type(y_w))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $g$:

```python
g=w*h+b*q3-q3+y_w
print('g=',g)
print('Type of g: ', type(g))
```

Основные функции, используемые в данной практической работе: `print(), type()`
