## Задание

Напишите функцию, которая по трем переданным ей числам определяет, могут ли эти числа быть сторонами одного и того же треугольника. Функция возвращает **True** или **False**.

## Решение

### Объявление функции

````python
def triangle(a, b, c):
  if a + b > c and a + c > b and b + c > a:
    return True
  else:
    return False
````

### Демонстрация использования функции

````python
a, b, c = 3, 4, 5
print(f'Стороны {a}, {b}, {c}: {triangle(a, b, c)}')

a, b, c = 1, 2, 4
print(f'Стороны {a}, {b}, {c}: {triangle(a, b, c)}')
````

Решите пример в соответствии с индивидуальным заданием

````python
#вариант 9
print(triangle(5, 6, 7))
````

1. Функция с возвратом значения возвращает результат с помощью `return`. Функция без возврата значения выполняет действия, но не возвращает результат.

2. Библиотечная функция — это готовая функция из библиотеки Python или подключенного модуля.

3. Библиотечные функции похожи на «черные ящики», потому что при их использовании не обязательно знать, как они устроены внутри. Достаточно знать, что передать функции и что она вернет.

4. `x = random.randint(1, 100)` записывает в `x` случайное целое число от 1 до 100 включительно.

5. `print(random.randint(1, 20))` выводит случайное целое число от 1 до 20 включительно.

6. `random.seed(20)` задает начальное значение генератора случайных чисел. При одинаковом значении `seed` последовательность случайных чисел будет одинаковой.

7. Функции позволяют вынести повторяющийся код в отдельный блок и вызывать его несколько раз.

8. Две части определения функции — заголовок и тело функции. В заголовке указываются имя и параметры, а в теле записываются выполняемые действия.

9. Параметры по умолчанию задаются при объявлении функции, например:

```python
def func(a=10):

```

10. Рекурсия — это способ, при котором функция вызывает саму себя.
